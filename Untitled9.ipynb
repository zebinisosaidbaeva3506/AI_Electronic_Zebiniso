```python
# Required libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

# 1. Create data
data = {
    'Income': [60000, 55000, 70000, 50000, 52000,
               48000, 45000, 40000, 30000, 35000,
               60000, 52000, 70000, 48000, 45000,
               50000, 51000, 49000, 53000, 55000],
    'Age':    [32, 35, 40, 30, 33,
               28, 25, 22, 27, 20,
               25, 28, 29, 30, 31,
               25, 29, 35, 31, 30],
    'Buy':    ['Yes', 'Yes', 'Yes', 'Yes', 'Yes',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'Yes', 'Yes']
}

# Create DataFrame
df = pd.DataFrame(data)

# 2. Print DataFrame
print("--- DataFrame Output ---")
print(df)
```

```python
# 3. Define Features (X) and Target (y)
X = df[['Income', 'Age']]
y = df['Buy']

# 4. Train Decision Tree Model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# 5. Visualize the Decision Tree
plt.figure(figsize=(12, 8))
tree.plot_tree(model,
               feature_names=['Income', 'Age'],
               class_names=['No', 'Yes'],
               filled=True)
plt.title("Decision Tree for 'Buy' Prediction")
plt.show()
```

```python
# 5️. Prediction
sample = pd.DataFrame({'Income':[ 47000, 51000], 'Age':[29, 31]})
pred = model.predict(sample)
print(pred)  # ['Yes', 'No']

####  Part 5 ###
# 6. Model rule print. not function f(x)
from sklearn.tree import export_text

# 트리의 규칙을 텍스트 형태로 출력
tree_rules = export_text(model, feature_names=['Income', 'Age'])
print(tree_rules)
```

```python
# Decision boundary
import numpy as np
import matplotlib.pyplot as plt

# 4. Scatter plot and decision boundary visualization
plt.figure(figsize=(10, 6))

# Create meshgrid for decision boundary
x_min, x_max = X['Income'].min() - 5000, X['Income'].max() + 5000
y_min, y_max = X['Age'].min() - 2, X['Age'].max() + 2

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 500),
    np.arange(y_min, y_max, 0.5)
)

# Predict for each point in the meshgrid
grid = np.c_[xx.ravel(), yy.ravel()]

# Use DataFrame to preserve feature names
grid_df = pd.DataFrame(grid, columns=['Income', 'Age'])

Z = model.predict(grid_df)

# Convert predictions into numerical values
Z = np.where(Z == 'Yes', 1, 0)
Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')

# Plot original data points
plt.scatter(
    X['Income'],
    X['Age'],
    c=np.where(y == 'Yes', 1, 0),
    cmap='coolwarm',
    edgecolors='black'
)

plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Decision Boundary - Decision Tree Classifier')
plt.colorbar(label='Predicted Class')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

# --- Data (read off your first plot) ---
data = [
    (70000, 40, 1), (55000, 35, 1), (52000, 33, 1), (60000, 32, 1),
    (53000, 31, 1), (50000, 30, 1), (55000, 30, 1),
    (49000, 35, 0), (45000, 31, 0), (48000, 30, 0), (48000, 28, 0),
    (51000, 29, 0), (52000, 28, 0), (70000, 29, 0), (30000, 27, 0),
    (45000, 25, 0), (50000, 25, 0), (60000, 25, 0), (40000, 22, 0),
    (35000, 20, 0),
]
df = pd.DataFrame(data, columns=["Income", "Age", "Buy"])
X, y = df[["Income", "Age"]], df["Buy"]

# --- Model ---
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# --- New data ---
new_income, new_age = 62000, 32
new_pred = model.predict(pd.DataFrame([[new_income, new_age]], columns=["Income", "Age"]))[0]
label = "Yes" if new_pred == 1 else "No"

# --- Decision boundary grid ---
x_min, x_max = X["Income"].min() - 5000, X["Income"].max() + 5000
y_min, y_max = X["Age"].min() - 2, X["Age"].max() + 1.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, 100),
                     np.arange(y_min, y_max, 0.1))
Z = model.predict(pd.DataFrame(np.c_[xx.ravel(), yy.ravel()],
                               columns=["Income", "Age"])).reshape(xx.shape)

# --- Plot ---
plt.figure(figsize=(10, 7))
plt.contourf(xx, yy, Z, alpha=0.2, colors=["red"])                 # red background
plt.contour(xx, yy, Z, colors="olive", alpha=0.4, linewidths=3)   # boundary lines

yes, no = df[df["Buy"] == 1], df[df["Buy"] == 0]
plt.scatter(yes["Income"], yes["Age"], c="green", marker="o", s=120, label="Buy: Yes")
plt.scatter(no["Income"], no["Age"], c="red", marker="x", s=100, label="Buy: No")
plt.scatter(new_income, new_age, c="blue", marker="*", s=350,
            label=f"New Data (Income: {new_income}, Age: {new_age} -> {label})")

plt.title("Decision Tree: Income vs Age with Decision Boundary & New Data Zebiniso")
plt.xlabel("Income")
plt.ylabel("Age")
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(True)
plt.legend(loc="upper left")
plt.show()
```
